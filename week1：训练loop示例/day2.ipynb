1.反向传播，但不涉及参数更新，只是记录梯度与参数的.grad中

```python
# arange()函数可以创建一个一维张量，包含从0到指定整数的所有整数（不包括该整数）
x = torch.arange(12)
y = torch.arange(12) * 2 + 1

# 初始化参数w,b,requires_grad=True表示在反向传播时需要计算梯度，pytorch会创建计算图记住w,b参与的计算
w = torch.tensor([0.0], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)
# loss关于w,b的梯度是通过loss.backward()计算得到的，
# pytorch会自动计算loss关于w,b的梯度并存储在w.grad和b.grad中

y_pred = w * x + b
loss = ((y_pred - y) ** 2).mean() # (y_pred - y) ** 2并不改变维度

print(f"loss: {loss}") # loss是一个标量张量，使用item()方法可以获取其数值
loss.backward() # 计算loss关于w,b的梯度，会更新w.grad和b.grad

w.grad, b.grad
```
