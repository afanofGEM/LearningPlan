```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
class MyLinear(nn.Module):
    def __init__(self,in_features,hidden_features,out_features):
        # 定义构造方法
        super().__init__() # 固定写法，让 PyTorch 正确管理模型

        self.net = nn.Sequential(
            nn.Linear(in_features,hidden_features),
            nn.ReLU(), # 激活函数让模型可以学习非线性函数，否则MLP就是Linear，非负数不变
            nn.Linear(hidden_features,out_features)
        )
    
    def forward(self,x):
        return self.net(x) # 直接调用Sequential对象就可以了
```

```python
x = torch.rand(200,4) # 200个样本，每个样本4个特征
y = torch.randint(0,4,(200,)) # 200个样本，每个样本是0-3之间的整数
y.shape # torch.Size([200])，每个样本是一个整数

moudle = MyLinear(in_features=4,hidden_features=5,out_features=4) # 4->5->1
opti = optim.Adam(moudle.parameters(),lr=0.01) # 用 Adam 优化器来更新 model 里面所有可训练参数，学习率是 0.01

for i in range(10):
    opti.zero_grad() # 梯度清零,清空每一个param.grad的值，否则会累加

    y_pred = moudle(x) # 前向传播
    loss_fn = nn.CrossEntropyLoss()
    loss = loss_fn(y_pred, y) # 分类问题计算交叉熵损失
    loss.backward() # 反向传播，计算梯度param.grad
    opti.step() # 通过param.grad更新参数
    print(f"epoch {i+1}, loss = {loss.item():.4f}")
```
